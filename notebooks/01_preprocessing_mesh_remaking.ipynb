```python
import napari
import trimesh

from morphotrack.meshes import remesh_uniform
```

```python
input_paths = [
    "../data/mesh_pia_original.ply",
    "../data/mesh_wm_original.ply",
]

### Uniform remeshing to ~target_area per face
target_area = 10000  # 100 micron x 100 micron

originals = [trimesh.load(p) for p in input_paths]
meshes = [remesh_uniform(m, target_area) for m in originals]
```

```python
colormaps = ["magenta", "green", "cyan", "yellow", "red", "blue"]

viewer = napari.Viewer(ndisplay=3)
for i, (mesh, original) in enumerate(zip(meshes, originals)):
    cmap = colormaps[i % len(colormaps)]
    viewer.add_surface((mesh.vertices, mesh.faces), colormap=cmap, blending="additive", shading="smooth")
    viewer.add_points(mesh.vertices, size=50, face_color=cmap, blending="additive")
    viewer.add_points(original.vertices, size=50, face_color="gray", blending="additive")
```

```python
for path, mesh in zip(input_paths, meshes):
    mesh.export(path.replace("_original", ""))
```
